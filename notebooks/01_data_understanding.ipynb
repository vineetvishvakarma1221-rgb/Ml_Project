### Date Columns identify 

```python
meta_col = ['CONS_NO','FLAG']
date_col = [col for col in df.columns if col not in meta_col]
print('date columns=',len(date_col))
print('first date=',date_col[0])
print('last date=',date_col[-1])
```

```python
date=pd.to_datetime(date_col,format='%m/%d/%Y')
print('minimun date=',date.min())
print('maximun date=',date.max())
print('Expected date count=',len(pd.date_range(date.min(),date.max(),freq='d')))
print('Actual date count=',len(date))
```

#### Consumption 

```python
consumption_data = df[date_col]
consumption_data.describe()
```

```python
print('Minimum',consumption_data.min().min())
print('Maximum',consumption_data.max().max())
print('Median',consumption_data.stack().median())
print('Mean',consumption_data.stack().mean())
```

#### Zero Consumption

```python
zero_cons=(consumption_data==0).sum().sum()
total_values=consumption_data.size
print('zero readings',zero_cons)
print('zero count percentage',abs(zero_cons/total_values)*100)
```
